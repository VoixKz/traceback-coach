```python
%load_ext traceback_coach
%coach_compact on        # fold the raw Python traceback into a collapsible, so the card leads
%coach_memory on

# Seed a realistic learner history so the memory review below has something to say.
from traceback_coach.hermes_memory import HermesMemory
_seed = HermesMemory()
_seed.forget()
for d in ["2026-07-08", "2026-07-09", "2026-07-11", "2026-07-12", "2026-07-14"]:
    _seed.record("IndexError", d)
_seed.record("KeyError", "2026-07-10"); _seed.record("KeyError", "2026-07-13")
_seed.record("NameError", "2026-07-12")
_seed.record_fixed("IndexError"); _seed.record_fixed("IndexError"); _seed.record_fixed("KeyError")
print("ready — run the two cells below and screenshot their output")
```

## 📸 Figure 1 — the coaching card (screenshot the output below)

```python
%%coach
# The coach turns the error into a guiding question + a diagram of the failing
# call graph. (%coach_compact folds the raw traceback into the small toggle.)
def third_item(seq):
    return seq[2]                       # fails when a row is too short

def middles(rows):
    return [third_item(r) for r in rows]

middles([[10, 20, 30], [40, 50]])       # IndexError on the 2nd row
```
