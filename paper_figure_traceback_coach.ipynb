## 📸 Figure 1 — the coaching workflow (full card + recursion call-graph)

```python
%%coach
# A recursive flatten with a bug three levels deep. The coach replaces the scary
# traceback with a guiding question and draws the RECURSION as a loop in the call graph.
def flatten(nested):
    out = []
    for item in nested:
        if isinstance(item, list):
            out += flatten(item)              # recursion
        else:
            out.append(item + 1)              # fails when item is a str
    return out

flatten([1, [2, [3, "x"]], 4])                # "x" + 1  ->  TypeError, deep in the recursion
```
