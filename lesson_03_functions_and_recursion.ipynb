## 1. A bug deep in a call chain (IndexError)

`run()` calls `average_of()` which calls `get_item()` — the bad index 5 starts in `run`.

```python
%%coach
def get_item(data, i):
    return data[i]
def average_of(data, indices):
    return sum(get_item(data, i) for i in indices) / len(indices)
def run():
    scores = [88, 92, 79]
    return average_of(scores, [0, 1, 5])
run()
```

Notice the diagram: the `5` enters in `run`, flows down to `get_item`, and breaks there.

## 2. Recursion with no stop (RecursionError)

`countdown` never reaches a base case.

```python
%%coach
def countdown(n):
    return countdown(n - 1)
countdown(5)
```
