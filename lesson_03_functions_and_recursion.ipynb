## 1 · 🔎 A bug deep in a call chain (IndexError)

`run()` calls `average_of()`, which calls `get_item()`. The bad index **5** enters the picture in `run` — but the crash only shows up inside `get_item`. Read the Coach's **diagram**: it traces the value from where it was created all the way down to where it breaks.

Run the cell and study how the diagram labels each step in the chain.

````python
%%coach
def get_item(data, i):
    return data[i]

def average_of(data, indices):
    return sum(get_item(data, i) for i in indices) / len(indices)

def run():
    scores = [88, 92, 79]          # only indices 0, 1, 2 are valid
    return average_of(scores, [0, 1, 5])   # 5 is out of range

run()
````

**Your turn:** look at the diagram — which function *introduces* the bad index, and which function *crashes* on it? Change `[0, 1, 5]` to `[0, 1, 2]` and re-run.

## 2 · 🔁 Recursion with no base case (RecursionError)

`countdown` keeps calling itself forever because there is no stopping condition. The Coach's diagram draws the function **looping back into itself** — so you can see at a glance that it never stops.

Run the cell and look at the diagram: notice the self-loop and the note about how many times it called itself.

````python
%%coach
def countdown(n):
    return countdown(n - 1)   # no base case — runs forever

countdown(5)
````

**Your turn:** add a base case so the recursion stops:

```python
def countdown(n):
    if n <= 0:
        return 0
    return countdown(n - 1)
```

Paste the fixed version into a new cell and re-run. What does the Coach say now?
