# 🧭 traceback-coach — Full Feature Tour

A **visual error-literacy coach** for Jupyter. When your code breaks, it doesn't hand you the fix — it shows you **why** it broke (with a diagram), and asks **one guiding question** so you learn to read errors yourself.

Run the cells top to bottom — each section shows off one feature. 👇

## 1. The anatomy card
Every error becomes a teaching card: a plain-language **translation**, a **diagram** of why it broke, **where** it happened, the **error family**, a *read-it-yourself* tip, a **worked example**, and **one question** — never the fix.

```python
%%coach
print(total)
```

## 2. The causal diagram — see where the bad value started
`run()` → `average_of()` → `get_item()`. The diagram traces the bad index **5** from where it entered all the way down to the line that breaks, with the code at each step.

```python
%%coach
def get_item(data, i):
    return data[i]
def average_of(data, indices):
    return sum(get_item(data, i) for i in indices) / len(indices)
def run():
    scores = [88, 92, 79]
    return average_of(scores, [0, 1, 5])   # index 5 doesn't exist
run()
```

## 3. 🔁 Recursion drawn as a loop
**Direct recursion** — the diagram shows the function with an arrow **into itself** (`calls itself ×N`), so it's instantly obvious it never stops.

```python
%%coach
def countdown(n):
    return countdown(n - 1)   # no base case
countdown(5)
```

**Mutual recursion** — two functions calling each other are drawn as a **cycle** (`loops back ×N`).

```python
%%coach
def is_even(n):
    return is_odd(n - 1)
def is_odd(n):
    return is_even(n - 1)
is_even(10)
```

## 4. 🌫️ Progressive fade — it teaches you to need it less
Run the **same** error three times: the 1st shows the full card, the 2nd a brief one, the 3rd just a minimal nudge — training you to read it yourself.

```python
%%coach
text = "hello"
text.push("!")   # strings have no .push
```

```python
%%coach
text = "hello"
text.push("!")   # 2nd time → brief
```

```python
%%coach
text = "hello"
text.push("!")   # 3rd time → minimal
```

## 8. 🧠 Personalized questions (LLM)
With an LLM configured, the guiding question references **your** actual variables. Compare a generic template vs a personalized question on the same bug:

```python
%coach_llm off
```

```python
%%coach
total = 0
for x in items:   # 'items' was never defined
    total += x
```

```python
%coach_llm on
```

```python
%%coach
total = 0
for x in items:   # now the question names 'items' and the loop
    total += x
```

## 9. 🪗 Compact mode — tame giant tracebacks

A RecursionError dumps **thousands** of repeated frames. Turn compact on, then run it **normally** (no `%%coach`) — Python's own traceback is folded to a short note instead of a wall of text. Turn it off to get the full traceback back.

```python
%coach_compact on
```

```python
def forever(n):
    return forever(n + 1)
forever(0)   # run normally — compact folds the traceback
```

```python
%coach_compact off
```

## 10. 👀 Watch mode — explain every cell automatically
Turn it on once, then just run normal cells (no `%%coach`). Fix a failing cell and the Coach celebrates with **✅ Fixed it!**

```python
%coach_watch on
```

```python
nums = [3, 6, 9]
print(nums[9])   # IndexError — auto-explained
```

Now fix it (a valid index) and re-run — watch for the **Fixed it!** note:

```python
nums = [3, 6, 9]
print(nums[2])   # fixed
```

```python
%coach_watch off
```
