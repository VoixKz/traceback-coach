## 1 · 📋 A position that doesn't exist (IndexError)

The list `scores` has **3 items**, so valid indices are `0`, `1`, and `2`.
Asking for `scores[3]` reaches past the end — Python raises an `IndexError`.

Run this cell and read the Coach's card: notice the error type, the line that broke, and the guiding question.

```python
%%coach
scores = [88, 92, 79]
print(scores[3])
```

**Your turn:** remember that indices start at `0` — the last valid index is always `len(scores) - 1`.
Change `3` to a valid index and re-run the cell.

## 2 · 🗝️ A key that isn't there (KeyError)

The dictionary `grades` was created with only one key: `"cs"`.
Asking for `grades["math"]` raises a `KeyError` because that key was never added.

Run this cell and read the card: pay attention to *which key* Python says is missing.

```python
%%coach
grades = {"cs": 90}
print(grades["math"])
```

**Your turn:** you have two options — either add `"math"` to the dictionary, or use
`grades.get("math")` which returns `None` safely instead of crashing.
Try both and re-run. What does the Coach's question point you toward?

---
Both errors share the same pattern: you asked for something that isn't there yet — always check *what's actually in* your list or dict before accessing it.
