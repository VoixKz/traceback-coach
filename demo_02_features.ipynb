# traceback-coach — features tour

This notebook showcases the key features of traceback-coach. A central insight:
when you re-run the **same error**, the coach message fades from full detail (with
diagram and worked example) → brief summary → minimal hint. You can force it back
with `%coach_level`, track what you've learned with `%coach_stats`, and keep learning
with `%coach_watch` and `%coach_lesson`.

```python
%load_ext traceback_coach
```

## Progressive fade — run the SAME error three times

```python
%%coach
print(total)
```

```python
%%coach
print(total)
```

```python
%%coach
print(total)
```
