# 🌟 traceback-coach — Features Tour

A hands-on tour of every major feature. Run it top-to-bottom (**Kernel → Restart Kernel and Run All Cells**), or step through with **Shift+Enter**.

> The Coach **never shows the fix** — it teaches you to read the error yourself.

**What to look for**
- 🌫️ progressive fade: the same error type fades full → brief → minimal as you repeat it
- 📊 error stats: a session summary of your most common mistakes
- 👀 watch mode: auto-explain every cell, then celebrate when you fix it
- 📖 lessons on demand: browse any error family without triggering it
- 🎯 quiz / active recall: guess the error type before the analysis reveals
- 🌏 bilingual: the whole Coach card in Traditional Chinese (zh-HK)
- 🪗 compact traceback: fold the red wall of text into one click-to-expand line

## 0 · Load the extension

```python
%load_ext traceback_coach
```

## 1 · 🌫️ Progressive fade

When you hit the **same error type** repeatedly, the Coach card fades from full detail → brief → minimal nudge — training you to need less scaffolding over time.

Switch to `auto` level, then run the same `ZeroDivisionError` **three times** (this error type appears **nowhere else** in this notebook, so the fade is clean).

```python
%coach_level auto
```

**Run 1 — full card** (diagram, translation, worked example, guiding question):

```python
%%coach
1 / 0   # first time → full card
```

**Run 2 — brief card** (shorter; diagram hidden):

```python
%%coach
1 / 0   # second time → brief
```

**Run 3 — minimal nudge** (just a one-line hint):

```python
%%coach
1 / 0   # third time → minimal
```

Restore full detail for all sections below:

```python
%coach_level full
```

## 2 · 📊 Error stats

`%coach_stats` shows a session summary of which error families you've hit most — great for spotting your weak spots.

```python
%coach_stats
```

## 3 · 👀 Watch mode + lessons on demand

**Watch mode** (`%coach_watch on`) auto-explains errors in every cell — no `%%coach` needed. When you fix the failing cell and re-run it, the Coach celebrates with **✅ Fixed it!**

```python
%coach_watch on
```

An `IndexError` — auto-explained because watch is on:

```python
xs = [10, 20, 30]
print(xs[9])   # IndexError — auto-explained
```

Now fix the index (change `9` → `2`) and re-run — look for the **Fixed it!** note. (If you're running all cells, the fixed version below shows it.)

```python
xs = [10, 20, 30]
print(xs[2])   # fixed — watch celebrates
```

```python
%coach_watch off
```

**Lessons on demand** — browse any error family's explanation, diagram, and worked example without having to trigger the error:

```python
%coach_lesson NameError
```

```python
%coach_lesson ZeroDivisionError
```

## 4 · 🎯 Quiz / active recall

With quiz mode on, the Coach hides its analysis behind a **"Reveal"** button. Predict the error type first, then click to check — recalling before seeing dramatically improves retention.

```python
%coach_quiz on
```

Guess the error type, then click **Reveal the Coach's analysis**:

```python
%%coach
scores = {"alice": 95, "bob": 88}
print(scores["charlie"])   # what type of error?
```

Turn quiz mode off to get instant feedback again:

```python
%coach_quiz off
```

## 6 · 🪗 Compact traceback

A `RecursionError` normally dumps **thousands** of repeated frames — a wall of red that buries the Coach card. Turn compact mode on, then run the error in a **plain cell** (no `%%coach`) to see the native Python traceback folded into one **click-to-expand** `<details>` line. The Coach card is completely untouched.

```python
%coach_watch on
%coach_compact on
```

Run a plain cell — the red traceback folds to one `▸ RecursionError … click to expand` line. Click it to see the full traceback. The 🧭 Coach card appears right below:

```python
def forever(n):
    return forever(n + 1)

forever(0)   # run normally — compact folds the traceback
```

Turn compact off to see the difference — full red traceback comes back:

```python
%coach_compact off
```

```python
def forever(n):
    return forever(n + 1)

forever(0)   # same error, full traceback now
```

```python
%coach_watch off
```
