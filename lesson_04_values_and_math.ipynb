## 1 · 🔤 Right type, wrong value — ValueError

`int()` is happy to convert a string to an integer — *as long as the string looks like a number*. Pass it the word `"ten"` and Python raises a `ValueError`: the type (`str`) was acceptable, but the value was not.

Run this cell and read the Coach's card: notice which value caused the problem and what `int()` actually requires.

```python
%%coach
n = int("ten")
print(n)
```

## 2 · ➗ Dividing by zero — ZeroDivisionError

Division by zero is undefined in mathematics, and Python enforces that strictly. Here `count` is `0` at the moment the division runs, so Python raises a `ZeroDivisionError` before it can compute `average`.

Run this cell and read the card: find the line that divides, and ask yourself why `count` is still `0` at that point.

```python
%%coach
count = 0
average = 100 / count
print(average)
```

**Your turn:** guard the division — check `if count != 0` before dividing, or set `count` to a non-zero value. Re-run the cell.

*Hint from the Coach card: what should happen when `count` is zero? Should the program skip the division, show a message, or use a default value?*
