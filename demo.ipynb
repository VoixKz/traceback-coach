## 2. Nested functions through an array (IndexError)

`run()` calls `average_of()`, which calls `get_item()`, which finally indexes
the list out of range. The diagram should trace the whole call chain down to
the line that breaks.

```python
%%coach
def get_item(data, i):
    return data[i]

def average_of(data, indices):
    total = 0
    for i in indices:
        total += get_item(data, i)
    return total / len(indices)

def run():
    scores = [88, 92, 79]
    return average_of(scores, [0, 1, 5])   # index 5 doesn't exist

run()
```

## 3. A type clash deep in a call chain (TypeError)

`build_label()` concatenates a string with a number several calls deep.

```python
%%coach
def build_label(name, score):
    return "score: " + score   # score is an int

def row(student):
    return build_label(student["name"], student["score"])

def table(students):
    return [row(s) for s in students]

table([{"name": "Ana", "score": 91}])
```

## 4. Direct recursion with no base case (RecursionError)

`factorial` calls itself forever. The diagram collapses the ~thousands of
identical frames into a single `factorial xN` node.

```python
%%coach
def factorial(n):
    return n * factorial(n - 1)   # missing base case!

factorial(5)
```

## 5. Mutual recursion — two functions calling each other (RecursionError)

`is_even` calls `is_odd` calls `is_even` … with no base case. The diagram shows
the first few alternating calls, an ellipsis for the thousands in between, and
the last few before the break.

```python
%%coach
def is_even(n):
    return is_odd(n - 1)

def is_odd(n):
    return is_even(n - 1)   # neither stops

is_even(10)
```
