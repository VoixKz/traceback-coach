# traceback-coach — demo

When a cell raises, the coach explains the traceback with a diagram, a worked
example, and one guiding question. It never shows the fix.

Run the cells top to bottom.

```python
%load_ext traceback_coach
```

## NameError — a name that was never defined

```python
%%coach
print(total)
```

## IndexError — a position that doesn't exist

```python
%%coach
xs = [1, 2, 3]
xs[5]
```

## TypeError — combining incompatible types

```python
%%coach
"5" + 1
```

## Open a lesson proactively (no error needed)

```python
%coach_lesson ZeroDivisionError
```
