# 🧭 traceback-coach — Guided Tour

A friendly walkthrough of the Coach in action. Each section triggers a real error
and shows you what the Coach produces — a plain-language card, a causal diagram,
and one guiding question that helps you understand the bug yourself.

Run it with **Kernel → Restart Kernel and Run All Cells**, or step through with
**Shift+Enter**.

> The Coach **never shows the fix** — it teaches you to read the error yourself.

**What to look for:**
- The colour-coded card that replaces the raw traceback
- Causal diagrams tracing exactly *where* a value went wrong
- Recursion drawn as a loop — even mutual recursion between two functions
- Watch mode auto-explaining every failing cell (no `%%coach` needed)
- A proactive lesson you can browse without triggering an error

## 2 · 📋 Nested functions through an array (IndexError)

`run()` calls `average_of()`, which calls `get_item()`, which indexes the list
out of range. Watch the diagram trace the bad index **5** all the way from
where it entered down to the line that actually breaks.


```python
%%coach
def get_item(data, i):
    return data[i]

def average_of(data, indices):
    total = 0
    for i in indices:
        total += get_item(data, i)
    return total / len(indices)

def run():
    scores = [88, 92, 79]
    return average_of(scores, [0, 1, 5])   # index 5 doesn't exist

run()
```

## 3 · 🔤 A type clash deep in a call chain (TypeError)

`build_label()` tries to concatenate a string with an integer several calls deep.
The diagram shows which step produced the bad type and where the crash happened.


```python
%%coach
def build_label(name, score):
    return "score: " + score   # score is an int, not a string

def row(student):
    return build_label(student["name"], student["score"])

def table(students):
    return [row(s) for s in students]

table([{"name": "Ana", "score": 91}])
```

## 4 · 🔁 Direct recursion — a function that calls itself forever (RecursionError)

Before the two recursion demos, force full detail so both diagrams render completely
even though they trigger the same error type twice.


```python
%coach_level full
```

`factorial` calls itself with no base case, so it recurses forever.
The diagram collapses the thousands of identical frames into a single
`factorial ×N` node with an arrow pointing back into itself — instantly
showing the loop.


```python
%%coach
def factorial(n):
    return n * factorial(n - 1)   # missing base case!

factorial(5)
```

## 5 · 🔄 Mutual recursion — two functions calling each other (RecursionError)

`is_even` calls `is_odd` calls `is_even` … with no base case in either.
The diagram shows the alternating calls as a **cycle** (`loops back ×N`),
making the infinite ping-pong obvious at a glance.


```python
%%coach
def is_even(n):
    return is_odd(n - 1)

def is_odd(n):
    return is_even(n - 1)   # neither function ever stops

is_even(10)
```

This cell has an out-of-range index. Watch the Coach card appear without `%%coach`.


```python
nums = [3, 6, 9]
print(nums[3])   # IndexError — index 3 doesn't exist in a 3-element list
```
