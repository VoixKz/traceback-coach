## 2 · 📊 Recursion as a loop — *mutual* recursion

`is_even` calls `is_odd` calls `is_even`… The diagram shows the two frames and a
dotted **`loops back ×N`** edge pointing back to the first — so it's obvious the
code cycles, not a 999-node chain.

```python
def is_even(n):
    return is_odd(n - 1)

def is_odd(n):
    return is_even(n - 1)

is_even(10)
```

## 3 · 📊 Nested calls → `IndexError`

Four functions calling each other. The diagram traces `run → average_of →
get_item → 💥 IndexError`, with the real code at each step.

```python
def get_item(data, i):
    return data[i]

def average_of(data, idxs):
    total = 0
    for i in idxs:
        total += get_item(data, i)
    return total / len(idxs)

def run():
    return average_of([10, 20, 30], [0, 1, 5])

run()
```
