# 🧭 traceback-coach — Full Feature Tour

A **visual error-literacy coach** for Jupyter. When your code breaks, the Coach shows you **why** it broke — anatomy card, causal diagram, guided question — so you understand the error, not just paste a fix.

**How to run:** **Kernel → Restart Kernel and Run All Cells**, or press **Shift+Enter** through it cell by cell.

> The Coach **never shows the fix** — it teaches you to read the error yourself.

**What to look for:**
- **Section 1** — the full anatomy card (translation · diagram · question)
- **Section 2** — causal diagram tracing a bad value across three functions
- **Sections 3–4** — recursion drawn as a self-loop and a two-function cycle
- **Section 5** — progressive fade: full card → brief → minimal, then reset
- **Section 6** — session error stats
- **Section 7** — quiz mode: guess the error type before the card reveals
- **Section 8** — bilingual: English then Traditional Chinese 繁體中文
- **Section 9** — LLM-personalized questions vs. offline template questions
- **Section 10** — compact mode: the red Python traceback folds into a single click-to-expand line
- **Section 11** — watch mode and the **✅ Fixed it!** celebration
- **Section 12** — browse any lesson without hitting the error
- **Section 13** — all commands at a glance

## 1 · 🗂️ The anatomy card

Every error becomes a teaching card with six parts:
a **plain-language translation**, a **causal diagram**, the **location** in your code, the **error family**, a *read-it-yourself* tip, a **worked example**, and **one guiding question** — never the fix.

The cell below triggers a `NameError` — watch how the Coach unpacks it.

```python
%%coach
print(total)
```

## 2 · 📐 Causal diagram — trace the bad value to its source

The diagram follows the bad index **5** from where it entered `run()` all the way down through `average_of()` to `get_item()`, showing the code at each hop. You can see at a glance *which call introduced the wrong value*.

```python
%coach_level full
```

```python
%%coach
def get_item(data, i):
    return data[i]

def average_of(data, indices):
    return sum(get_item(data, i) for i in indices) / len(indices)

def run():
    scores = [88, 92, 79]
    return average_of(scores, [0, 1, 5])   # index 5 doesn't exist

run()
```

## 3 · 🔁 Recursion drawn as a self-loop

Direct recursion with no base case produces thousands of identical frames. The Coach collapses them into a **single node with an arrow back into itself** (`calls itself ×N`) — so the infinite loop is immediately obvious.

```python
%%coach
def countdown(n):
    return countdown(n - 1)   # no base case — calls itself forever

countdown(5)
```

## 4 · 🔄 Mutual recursion drawn as a cycle

Two functions that call each other with no exit condition. The diagram shows a **two-node cycle** (`loops back ×N`) so the bouncing is plain to see.

```python
%%coach
def is_even(n):
    return is_odd(n - 1)

def is_odd(n):
    return is_even(n - 1)   # neither function ever stops

is_even(10)
```

## 5 · 🌫️ Progressive fade — learn to need it less

Run the **same error type three times**. The Coach fades its detail level automatically: **full card** the first time, a **brief version** the second, a **minimal nudge** the third — training you to read errors independently.

The `AttributeError` below is used **only in this section**, so the full → brief → minimal progression is clean.

First encounter — full card:

```python
%%coach
text = "hello"
text.push("!")   # strings have no .push
```

Second encounter — brief card:

```python
%%coach
text = "hello"
text.push("!")   # 2nd time → brief
```

Third encounter — minimal nudge:

```python
%%coach
text = "hello"
text.push("!")   # 3rd time → minimal
```

Force full detail back for the rest of the tour:

## 9 · 🧠 Personalized questions (LLM)

With an LLM configured, the guiding question references **your actual variable names**. Compare the generic template question (LLM off) with a personalized one (LLM on) for the exact same bug.

```python
%coach_llm off
```

Offline template question — generic wording:

```python
%%coach
total = 0
for x in items:   # 'items' was never defined
    total += x
```

```python
%coach_llm on
```

LLM-personalized question — names `items` and the loop context:

```python
%%coach
total = 0
for x in items:   # now the question can name 'items' and the loop
    total += x
```

## 10 · 🪗 Compact mode — fold the red traceback into one line

A `RecursionError` normally dumps **thousands of repeated frames** — a wall of red text. With compact mode on, the Python traceback is folded into a single **▸ click-to-expand `<details>` line**. The Coach card itself is unaffected.

Compact mode only fires on **uncaught** errors (plain cells or watch mode), not inside `%%coach` cells — because `%%coach` catches the exception before the custom handler runs.

Below we use `%coach_watch on` so you also get the Coach card *alongside* the folded traceback, showing both features together.

```python
%coach_compact on
```

```python
%coach_watch on
```

Run this cell — Python's red traceback folds into one click-to-expand line, and the Coach card appears above it (because watch mode is on).

```python
def forever(n):
    return forever(n + 1)

forever(0)   # uncaught — compact folds the traceback, watch shows the card
```

```python
%coach_watch off
```

```python
%coach_compact off
```

## 11 · 👀 Watch mode — explain every cell automatically

Turn watch mode on once and the Coach explains **any failing plain cell** automatically — no `%%coach` needed. Fix the broken cell and re-run it: the Coach celebrates with a green **✅ Fixed it!** banner.

```python
%coach_watch on
```

This `IndexError` will be explained automatically:

```python
nums = [3, 6, 9]
print(nums[9])   # IndexError — auto-explained by watch mode
```

Now fix it (valid index) and re-run the cell — watch for the **Fixed it!** banner:

```python
nums = [3, 6, 9]
print(nums[2])   # fixed — Coach shows ✅ Fixed it!
```

```python
%coach_watch off
```
